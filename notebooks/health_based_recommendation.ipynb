```python
import os
import sys
import pandas as pd

# Permette di trovare config.py facendo un passo indietro nella cartella principale
sys.path.append(os.path.abspath(os.path.join('..')))

from config import PATH_CLEAN_RECIPES
from models.health_based import HealthBasedRecommender

# CARICAMENTO DEL DATASET (Questo crea la variabile df_recipes)
df_recipes = pd.read_csv(os.path.join('..', PATH_CLEAN_RECIPES))

print(f"Dataset caricato correttamente! Numero di ricette: {len(df_recipes)}")
```

```python
from models.health_based import HealthBasedRecommender

# Inizializziamo il modello salutistico
health_model = HealthBasedRecommender()
health_model.fit(df_recipes)

# TEST 1: Ricerca singola con vincoli rigidi e profilo "Perdita Peso" (Weight Loss)
print("--- TEST 1: TOP 3 RICETTE VEGETARIANE, IPOCALORICHE, AD ALTO APPORTO PROTEICO ---")
ricette_sane = health_model.recommend(
    max_calories=500, 
    min_protein_pct=20,          # Almeno il 20% del fabbisogno giornaliero di proteine
    tags_required=['vegetarian'], 
    profile_name='weight_loss', 
    top_k=3
)

for r in ricette_sane:
    print(f"\nNome: {r['name'].upper()}")
    print(f"   Score Profilo: {r['health_score']} | Calorie: {r['calories']} kcal | Proteine: {r['protein_pdv']}% PDV")


# TEST 2: Generazione Piano Settimanale di 7 giorni con profilo "Bilanciato" e Gluten-Free
print("\n" + "="*60 + "\n")
print("--- TEST 2: PIANO ALIMENTARE SETTIMANALE (GLUTEN-FREE & BILANCIATO) ---")
piano_settimanale = health_model.generate_weekly_plan(
    max_calories=600, 
    tags_required=['gluten-free'], 
    profile_name='balanced', 
    days=7
)

for giorno, r in piano_settimanale.items():
    print(f"📅 {giorno}: {r['name'].upper()} ({r['calories']} kcal, Proteine: {r['protein_pdv']}%, Grassi: {r['fat_pdv']}%)")
```
