```python
import os
import sys
import pandas as pd

# Permette di trovare config.py facendo un passo indietro nella cartella principale
sys.path.append(os.path.abspath(os.path.join('..')))

from config import PATH_CLEAN_RECIPES
from models.mood_based import MoodBasedRecommender

# CARICAMENTO DEL DATASET (Questo crea la variabile df_recipes)
df_recipes = pd.read_csv(os.path.join('..', PATH_CLEAN_RECIPES))

print(f"Dataset caricato correttamente! Numero di ricette: {len(df_recipes)}")
```

```python
from models.mood_based import MoodBasedRecommender

# Inizializziamo il recommender basato sul Mood
mood_model = MoodBasedRecommender()
mood_model.fit(df_recipes)

# SCENARIO A: L'utente è stanco, stressato, vuole uno sgarro pesante, confortevole, super saporito e veloce da fare
# Configurazione: Body+, Time- (veloce), Taste+, Price- (economico), Mental+ (comfort)
print("--- SCENARIO A: COMFORT FOOD VELOCE E RICCO DI SAPORE (STRESS DA ESAME) ---")
risultati_comfort = mood_model.recommend(
    body=4.0, time=-4.0, taste=4.0, price=-3.0, mental=4.0, modification=-2.0, 
    top_k=3
)
for r in risultati_comfort:
    print(f"\nRicetta: {r['name'].upper()} (Distanza: {r['distanza_geometrica']})")
    print(f"   Dettagli: {r['calorie']} kcal | Tempo: {r['minuti']} min")
    print(f"   Vettore Ricetta: {r['mood_scores']}")


# SCENARIO B: L'utente si sente in colpa, vuole mangiare super leggero, salutare, ha tempo per cucinare qualcosa di sofisticato/classico
# Configurazione: Body- (leggero), Time+ (elaborato), Taste- (delicato), Mental- (salutare)
print("\n" + "="*60 + "\n")
print("--- SCENARIO B: PIATTO LEGGERO, SALUTARE ED ELABORATO (DETOX DOMENICALE) ---")
risultati_detox = mood_model.recommend(
    body=-4.0, time=4.0, taste=-4.0, price=2.0, mental=-4.0, modification=-3.0, 
    top_k=3
)
for r in risultati_detox:
    print(f"\nRicetta: {r['name'].upper()} (Distanza: {r['distanza_geometrica']})")
    print(f"   Dettagli: {r['calorie']} kcal | Tempo: {r['minuti']} min")
    print(f"   Vettore Ricetta: {r['mood_scores']}")
```
