```python
import os
import sys
import pandas as pd

# Permette di trovare config.py facendo un passo indietro nella cartella principale
sys.path.append(os.path.abspath(os.path.join('..')))

from config import PATH_CLEAN_RECIPES
from models.content_based import ContentBasedRecommender

# CARICAMENTO DEL DATASET (Questo crea la variabile df_recipes)
df_recipes = pd.read_csv(os.path.join('..', PATH_CLEAN_RECIPES))

print(f"Dataset caricato correttamente! Numero di ricette: {len(df_recipes)}")
```

```python
from models.content_based import ContentBasedRecommender

# Inizializziamo il recommender indicando dove salvare la matrice sparsa
content_model = ContentBasedRecommender(matrix_path="../dataset/tfidf_matrix.npz")

# Addestramento (Fase Offline)
content_model.fit(df_recipes)

# Simuliamo uno Svuota-Frigo: l'utente ha questi ingredienti a casa
ingredienti_frigo = ["chicken breast", "lemon", "garlic", "rosemary"]

# Chiediamo le top 5 ricette che richiedono al massimo 2 ingredienti mancanti
print(f"--- SVUOTA-FRIGO: Risultati per {ingredienti_frigo} ---")
risultati_frigo = content_model.recommend(ingredienti_frigo, max_missing_ingredients=2, top_k=5)

for i, r in enumerate(risultati_frigo, 1):
    print(f"\n{i}. [{r['similarity']}] {r['name'].upper()}")
    print(f"   Possiedi già: {r['ingredienti_trovati']}")
    print(f"   Ti mancano: {r['ingredienti_mancanti']}")
    print(f"   Calorie: {r['calorie']} kcal | Tempo: {r['minuti']} min")
```
